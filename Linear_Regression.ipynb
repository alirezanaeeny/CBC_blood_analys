```
import pandas as pd
import sklearn
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

df = pd.read_csv("CBC_blood_analys.csv")
X = df[["hb"]]
Y = df["hct"]

X_train, X_test, Y_train, Y_test = train_test_split(
    X, Y, test_size=0.2, random_state=42
)

model = LinearRegression()
model.fit(X_train, Y_train)

print("slope:", model.coef_)
print("intercept:", model.intercept_)

Y_pred = model.predict(X_test)
r2 = r2_score(Y_test, Y_pred)
print("r2_score:", r2)

df2=pd.read_csv("CBC2.csv",encoding="cp1256",sep=";",
usecols=['CBC,Diff-CBC & Diff-Hb',"CBC,Diff-CBC & Diff-R.B.C",
 'CBC,Diff-CBC & Diff-Hct',"محل درخواست"])
df2=df2.rename(columns={'CBC,Diff-CBC & Diff-Hb':"hb", 'CBC,Diff-CBC & Diff-Hct':"hct",
"محل درخواست":"ward","CBC,Diff-CBC & Diff-R.B.C":"rbc"})
X_new=df2[["hb"]]
Y_new_pred=model.predict(X_new)
Y_actual=df2["hct"]
r2_new=r2_score(Y_actual,Y_new_pred)
print("r2_new_score:",r2_new)
df2["new_hct"]=Y_new_pred
print(df2[["hct","new_hct"]])

X_multi=df[["hb","rbc"]]
Y_multi=df["hct"]
X_train_m,X_test_m,Y_train_m,Y_test_m=train_test_split(X_multi,Y_multi,
test_size=0.2,random_state=42)
model_multi=LinearRegression()
model_multi.fit(X_train_m,Y_train_m)
print("slope_m :",model_multi.coef_)
print("intercept_m :",model_multi.intercept_)
Y_pred_m=model_multi.predict(X_test_m)
r2_m=r2_score(Y_test_m,Y_pred_m)
print("r2_score_m :",r2_m)

X_multi_n=df2[["hb","rbc"]]
Y_pred_m_n=model_multi.predict(X_multi_n)
df2["new_hct_m"]=Y_pred_m_n
r2_m_n=r2_score(Y_actual,Y_pred_m_n)
print("r2_m_n",r2_m_n)
print(df2[["hct","new_hct","new_hct_m"]])
```
